```python
import pandas as pd

dados = {"Nome": ["Ana", "João", "Maria"], "idade": [20,25,30]}
df = pd.DataFrame(dados)
df
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.plot(
    df["Nome"],
    df["idade"]
)

plt.title("Idade dos Participantes")
plt.xlabel("Nome")
plt.ylabel("Idade")

plt.show()
```
